## Data Type Validation

Review the assigned data types and convert columns to types that more accurately represent their values and intended use.

```python
# Review the distinct values of columns intended for the category data type to verify that their values are appropriate for categorical conversion.
category_columns = [
    'Gender', 
    'Offer', 
    'Internet Type',
    'Contract', 
    'Payment Method', 
    'Customer Status', 
    'Churn Category', 
    'Churn Reason'
]


for column in category_columns:
    print(churn_df[column].value_counts(dropna=False), '\n\n*===============*')
```

```python
# Review the distinct values of columns intended for the boolean data type to verify that their values are appropriate for boolean conversion.
boolean_columns = [
    'Married',
    'Phone Service',
    'Multiple Lines',
    'Internet Service',
    'Online Security',
    'Online Backup',
    'Device Protection Plan',
    'Premium Tech Support',
    'Streaming TV',
    'Streaming Movies',
    'Streaming Music',
    'Unlimited Data',
    'Paperless Billing'
]

for column in boolean_columns:
    print(churn_df[column].value_counts(dropna=False), '\n\n*===============*')
```

```python
bool_mapping = {
    'Yes': True,
    'No': False
}

for column in boolean_columns:
    churn_df[column] = churn_df[column].map(bool_mapping)
```

```python
# Group columns by their intended data type and convert
int8_columns = [
    'Age',
    'Number of Dependents',
    'Number of Referrals',
    'Tenure in Months'
]

float32_columns = [
    'Latitude',
    'Longitude',
    'Avg Monthly Long Distance Charges',
    'Avg Monthly GB Download',
    'Monthly Charge',
    'Total Charges',
    'Total Refunds',
    'Total Long Distance Charges',
    'Total Revenue'
]


dtype_mapping = {
    **dict.fromkeys(int8_columns, 'Int8'),
    **dict.fromkeys(float32_columns, 'float32'),
    **dict.fromkeys(category_columns, 'category'),
    **dict.fromkeys(boolean_columns, 'boolean'),
    'Zip Code': 'string',
    'Total Extra Data Charges': 'Int16'
}

churn_df = churn_df.astype(dtype_mapping)
```

```python
# Review the resulting data types to confirm that the conversions were applied successfully
churn_df.info()
```

## Null Values Assessment

Identify columns containing missing values and determine whether the nulls represent data quality issues or expected structural values.

```python
churn_df.isna().sum()
```

### Offer

According to the telecom data dictionary, `Offer` represents the last marketing offer accepted by the customer. Valid values are `None` and `Offer A` through `Offer E`.

Initial inspection shows that the dataset contains `Offer A` through `Offer E` and null values, but no explicit `None` values.

```python
# Check for explicit 'None' values in the Offer column
churn_df[churn_df['Offer'] == 'None']
```

```python
# Add 'None' as a category in the Offer column and set null values in the column to it
churn_df['Offer'] = (
    churn_df['Offer']
    .cat.add_categories('None')
    .fillna('None')
)

churn_df['Offer'].value_counts(dropna=False)
```

No explicit `None` values were found. Based on the data dictionary and the observed values, the null values are interpreted as customers who did not accept a marketing offer and are replaced with `None` to make this meaning explicit.

The `Avg Monthly Long Distance Charges` and `Multiple Lines` columns contain the same number of null values. Since both fields depend on whether a customer has phone service, the null values may correspond to customers where `Phone Service` is `False`, indicating that they did not subscribe to the telco's phone service. Further investigation is needed to verify this relationship.

```python
# Check for customers with 0 average monthly long-distance charges
churn_df[churn_df['Avg Monthly Long Distance Charges'] == 0]
```

```python
churn_df['Phone Service'].value_counts(dropna=False)
```

```python
churn_df['Multiple Lines'].value_counts(dropna=False)
```

```python
# Since there are no 0 values in 'Avg Monthly Long Distance Charges', investigate whether its null values correspond to customers without phone service
# and whose 'Multiple Lines' value is also null

churn_df[(churn_df['Phone Service'] == False) & 
         (churn_df['Avg Monthly Long Distance Charges'].isnull()) & 
         (churn_df['Multiple Lines'].isnull())].shape[0]
```

```python
# Impute structural nulls of 'Avg Monthly Long Distance Charges' and 'Multiple Lines' to explicitly represent customers without phone service

churn_df = churn_df.fillna(
    {
        'Avg Monthly Long Distance Charges': 0,
        'Multiple Lines': False
    }
)
```

```python
print(churn_df['Avg Monthly Long Distance Charges'].isna().sum())
print(churn_df['Multiple Lines'].isna().sum())
```

### Internet Service-Related Missing Values

The following columns contain the same number of null values: `Internet Type`, `Avg Monthly GB Download`, `Online Security`, `Online Backup`, `Device Protection Plan`, `Premium Tech Support`, `Streaming TV`, `Streaming Movies`, `Streaming Music`, and `Unlimited Data`.

Since these fields are dependent on whether a customer has internet service, the null values are investigated to determine whether they correspond to customers without an internet service subscription.

```python
# Check whether all internet-related null values correspond to customers without internet service
conditions = (
    churn_df['Internet Service'] == False &
    churn_df['Internet Type'].isna() &
    churn_df['Avg Monthly GB Download'].isna() &
    churn_df['Online Security'].isna() &
    churn_df['Online Backup'].isna() &
    churn_df['Device Protection Plan'].isna() &
    churn_df['Premium Tech Support'].isna() &
    churn_df['Streaming TV'].isna() &
    churn_df['Streaming Movies'].isna() &
    churn_df['Streaming Music'].isna() &
    churn_df['Unlimited Data'].isna()
)

churn_df[conditions].shape[0]
```

```python
# Replace structural nulls with values representing the absence of internet-related services

churn_df['Internet Type'] = churn_df['Internet Type'].cat.add_categories('None')

churn_df = churn_df.fillna(
    {
        'Internet Type': 'None',
        'Avg Monthly GB Download': 0,
        'Online Security': False,
        'Online Backup': False,
        'Device Protection Plan': False,
        'Premium Tech Support': False,
        'Streaming TV': False,
        'Streaming Movies': False,
        'Streaming Music': False,
        'Unlimited Data': False

    }
)
```

### Churn-Related Missing Values

`Churn Category` and `Churn Reason` contain the same number of null values. Since these fields describe why a customer churned, their availability is expected to depend on `Customer Status`.


```python
# Investigate null values to determine whether they correspond to customers who have not churned

conditions = (
    (churn_df['Customer Status'] != 'Churned') &
    (churn_df['Churn Category'].isna()) &
    (churn_df['Churn Reason'].isna())
)

churn_df[conditions].shape[0]
```

The null values in `Churn Category` and `Churn Reason` correspond to customers who have not churned. Since these fields are not applicable to non-churned customers, the null values are retained rather than imputed to avoid introducing artificial or misleading information into the dataset.

## Zip Code Population Data Validation

Validate the supporting zip code population dataset.

```python
population_df = pd.read_csv('../data/raw/telecom_zipcode_population.csv')
```

```python
population_df.info()
```

```python
# Although Zip Code contains numeric values, it is best to convert it to a string as it functions as an identifier rather than a quantitative variable.
population_df['Zip Code'] = population_df['Zip Code'].astype('string')
```

```python
# Check for duplicates
print(population_df.duplicated().sum())
print(population_df['Zip Code'].duplicated().sum())
```

```python
population_df['Population'].describe()
```

### Population Validation

The minimum population value in the dataset is 11. Since this value appears unusually low for a ZIP code, records with very low population estimates are investigated to determine whether they represent valid observations or potential data quality issues.

```python
population_df[population_df['Population'] < 50]
```

Although some values were reviewed as potential anomalies, small ZIP code populations may occur depending on the geographic area and population estimation methodology.

Since the data dictionary defines these values as current population estimates and provides no reference source or year for independent validation, the original population values are retained rather than applying unsupported corrections.

```python
# Check whether all customer Zip Codes have a corresponding record in the population dataset
churn_df.loc[
    ~churn_df['Zip Code'].isin(population_df['Zip Code']),
    'Zip Code'
].unique()
```

### Zip Code Population Validation Summary

The supporting ZIP code population dataset contains no missing values or duplicate ZIP codes, and all population values are positive. Unusually small population estimates were reviewed and retained because no reference source or year was provided for correction.

All ZIP codes in the customer dataset have a corresponding record in the population dataset, confirming complete referential coverage between the two datasets.

```python
# Export to Parquet format for use in subsequent exploratory analysis notebooks.
population_df.to_parquet('../data/processed/telecom_zipcode_population_clean.parquet', index=False)
```
